```python
# PROGRAM 6:
# Write a program using image processing techniques
# to detect traffic signs based on their color
# from an input image.

import cv2
import numpy as np
import matplotlib.pyplot as plt


# --------------------------------------------------
# 1. Read the input image
# --------------------------------------------------

image = cv2.imread("traffic_signs.jpeg")

if image is None:
    raise FileNotFoundError(
        "traffic_signs.jpeg not found"
    )


# --------------------------------------------------
# 2. Convert image from BGR to HSV
# --------------------------------------------------

hsv = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2HSV
)


# --------------------------------------------------
# 3. Detect RED color
# --------------------------------------------------

red1 = cv2.inRange(
    hsv,
    (0, 80, 80),
    (10, 255, 255)
)

red2 = cv2.inRange(
    hsv,
    (170, 80, 80),
    (180, 255, 255)
)

red = red1 | red2


# --------------------------------------------------
# 4. Detect BLUE color
# --------------------------------------------------

blue = cv2.inRange(
    hsv,
    (90, 80, 70),
    (130, 255, 255)
)


# --------------------------------------------------
# 5. Detect YELLOW color
# --------------------------------------------------

yellow = cv2.inRange(
    hsv,
    (18, 80, 80),
    (40, 255, 255)
)


# --------------------------------------------------
# 6. Combine all color masks
# --------------------------------------------------

combined = red | blue | yellow


# --------------------------------------------------
# 7. Remove noise using morphological operations
# --------------------------------------------------

kernel = np.ones(
    (5, 5),
    np.uint8
)

combined = cv2.morphologyEx(
    combined,
    cv2.MORPH_OPEN,
    kernel
)

combined = cv2.morphologyEx(
    combined,
    cv2.MORPH_CLOSE,
    kernel
)


# --------------------------------------------------
# 8. Find contours
# --------------------------------------------------

contours, _ = cv2.findContours(
    combined,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)


# --------------------------------------------------
# 9. Detect traffic signs
# --------------------------------------------------

for cnt in contours:

    # Calculate area
    area = cv2.contourArea(cnt)

    # Ignore small objects
    if area < 150:
        continue

    # Calculate perimeter
    perimeter = cv2.arcLength(
        cnt,
        True
    )

    if perimeter == 0:
        continue

    # Calculate circularity
    circularity = (
        4 * np.pi * area
        / (perimeter * perimeter)
    )

    # Check whether contour is circular
    if circularity > 0.35:

        # Get bounding rectangle
        x, y, w, h = cv2.boundingRect(
            cnt
        )

        # Draw bounding box
        cv2.rectangle(
            image,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )

        # Add label
        cv2.putText(
            image,
            "Traffic Sign",
            (x, max(y - 8, 20)),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (0, 255, 255),
            2
        )


# --------------------------------------------------
# 10. Convert BGR to RGB for Matplotlib
# --------------------------------------------------

image_rgb = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2RGB
)


# --------------------------------------------------
# 11. Display result in JupyterLab
# --------------------------------------------------

plt.figure(
    figsize=(12, 8)
)

plt.imshow(
    image_rgb
)

plt.title(
    "Color-Based Traffic Sign Detection"
)

plt.axis("off")

plt.show()
```
